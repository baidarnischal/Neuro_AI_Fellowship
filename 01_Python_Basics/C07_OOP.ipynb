```python
class Car:
    def __init__(self, name, model, year, top_speed, price):
        self.name = name
        self.model = model
        self.year = year
        self.top_speed = top_speed
        self.price = price

    def display(self):
        print(f"Car name is {self.name} and model {self.model}")
        print(f"Top_speed = {self.top_speed} and it costs {self.price}")

obj = Car("BMW", "M3", 2027, 300,7000000)
obj.display()
```
