### Question: Write a program that displays the outputs of AND, OR, NAND, NOR, XOR, and XNOR for all four possible combinations of two binary inputs.

```python
a = 1
b = 0

a_and_b = a & b
print(f"AND: {a_and_b}")

a_or_b = a | b
print(f"OR: {a_or_b}")

a_xor_b = a ^ b
print(f"XOR: {a_xor_b}")

a_nand_b = int(not (a & b))
print(f"NAND: {a_nand_b}")

a_nor_b = int(not (a | b))
print(f"NOR: {a_nor_b}")

a_xnor_b = int(not (a ^ b))
print(f"XNOR: {a_xnor_b}")
```
