if else statement

```python
### WAP to find whether a number is even or odd.

number = 3

if number%2 == 0:
    print(f"{number} is an even number")

else:
    print(f"{number} is an odd number")
```

```python
### WAP to find whether a number is positive, negative or 0

number = 10

if number>0:
    print("Positive")

elif number<0:
    print("Negative")

else:
    print("zero")
```

```python
marks = 100

if marks < 0 or marks > 100:
    print("Invalid should be between 0-100")

elif marks >= 80:
    print("Distinction")

elif 70 <= marks < 80:
    print("Second Division")

elif 50 <= marks < 70:
    print("Third Division")

elif 40 <= marks < 50:
    print("Pass")

else:
    print("Fail")
```
