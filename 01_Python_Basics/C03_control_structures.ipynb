if else statement

```python
### WAP to find whether a number is even or odd.

number = 4

if number%2 == 0:
    print(f"{number} is an even number")

else:
    print(f"{number} is an odd number")
```

```python
### WAP to find whether a number is positive, negative or 0

number = 10

if number>0:
    print("Positive")

elif number<0:
    print("Negative")

else:
    print("zero")
```

```python
marks = 100

if marks < 0 or marks > 100:
    print("Invalid should be between 0-100")

elif marks >= 80:
    print("Distinction")

elif 70 <= marks < 80:
    print("Second Division")

elif 50 <= marks < 70:
    print("Third Division")

elif 40 <= marks < 50:
    print("Pass")

else:
    print("Fail")
```

lambda function

```python
var = lambda x,y:x+y
var(2,2)
```

```python
even_odd = lambda x:print(f"{x} is even") if x%2==0 else print("odd")

even_odd(3)
```

```python
even_odd = (
    lambda x:
    print(f"{x} is even") 
    if x%2==0 
    else 
    print("odd")
    )

even_odd(3)
```

```python
positive_negative_zero = lambda x: print(f"{x} is a positive number") if x>0 else print(f"{x} is a negative number") if x<0 else print(f"{x} is zero")

positive_negative_zero(0)
positive_negative_zero(10)
positive_negative_zero(-1)
```

```python
check_marks = lambda marks: (
    print("Invalid") if marks < 0 or marks > 100
    else print("Distinction") if marks >= 80
    else print("First Division") if marks >= 70
    else print("Second Division") if marks >= 60
    else print("Third Division") if marks >= 50
    else print("Fourth Division") if marks >= 40
    else print("Fail")
)

check_marks(55)
```
