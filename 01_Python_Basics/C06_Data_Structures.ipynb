```python
# Armstrong number

mylist = [123, 534, 153, 1634]

for item in mylist:
    length_of_item = 0
    original_digit = item
    sum = 0

    # COunt num of digits
    while item != 0:
        item = item //10
        length_of_item+=1

    item = original_digit

    # Calculate Armstrong sum
    while item!=0:
        last_digit = item%10
        sum = sum + last_digit**length_of_item
        item = item // 10

    if sum == original_digit:
        print(original_digit, "is an Armstrong number")
    else:
        print(original_digit, "is not an Armstrong number")
```
