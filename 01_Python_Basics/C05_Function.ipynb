```python
# Using Function to calculate square of given number

def square(n):
    return n**2

number = int(input("Enter a number"))

square(number)
```

```python
# WAP to calculate the factorial of a number using for loop

number = int(input("Enter a number"))

fact = 1

for i in range(1,number+1):
    fact*=i

print(f"{number}!={fact}")
```

```python
# WAP to calculate the factorial of a number using function

number = int(input("Enter a number"))


def factorial(number):
    fact = 1
    for i in range(1,number+1):
        fact*=i
    return fact

    
print(f"{number}!={factorial(number)}")
```

```python
# WAP to calculate the factorial of a number using recursive function

number = int(input("Enter a number"))


def factorial(n):
    if n<0:
        print(f"{n} is negative. Factorial doesn't exist.")
    elif n == 0 or n == 1:
        return 1
    else:
        return n*factorial(n-1) 

    
print(f"{number}!={factorial(number)}")
```

### WAP to calculate fibonacci series using: for loop, while loop, function, and recursive function

#### for loop

```python
a = 0
b = 1
n = 10

for i in range(n):
    print(a, end = " ")
    c = a+b
    a = b
    b = c
```

#### while loop

```python
a = 0
b = 1
n = 10
i = 1

while(i<=n):
    print(a, end = " ")
    c = a+b
    a = b
    b = c
    i+=1
```

#### function

```python
def getfibonacci(n):
    a = 0
    b = 1
    for i in range(n):
        print(a, end = " ")
        c = a+b
        a = b
        b = c

n = 10
getfibonacci(n)
```

#### recursive function

```python
def fibonacci(a, b, n):
    if n == 0:
        return

    print(a, end=" ")
    fibonacci(b, a + b, n - 1)


fibonacci(0, 1, 10)
```

```python
def fibonacci(n):
    if n <= 1:
        return n

    return fibonacci(n-1) + fibonacci(n-2)


for i in range(10):
    print(fibonacci(i), end=" ")
```
