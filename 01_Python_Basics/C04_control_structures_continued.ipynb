```python
# WAP to find the factorial of a number using for loop

number = 6
factorial = 1

for i in range(1,number+1):
    factorial *=i

print(factorial)
```

```python
# WAP to find the factorial of a number using while loop

number = 5
factorial = 1
i = 1

while i<=number:
    factorial*=i
    i+=1

print(factorial)
```

```python
# WAP to reverse a number using while loop

number = 123143
reverse = 0

while number > 0:
    last_digit = number % 10
    reverse = reverse*10 + last_digit
    number = number//10

print(reverse)
    

```

```python
# WAP to reverse a number using for loop

number = 123143
reverse = 0


for i in range(len(str(number))):
    last_digit = number % 10
    reverse = reverse*10 + last_digit
    number = number//10

print(reverse)
    

```

```python
# WAP to find sum of each digit of a number using while loop

number = 123124
sum = 0

while number > 0:
    last_digit = number % 10
    sum+=last_digit
    number = number//10

print(sum)
```

```python
# WAP to print all numbers from 1 to 100 using while loop
i = 1
while i<=100:
    print(i)
    i+=1
```

```python
# WAP to print even numbers from 1 to 100 using while loop
i = 1
while i<=100:
    if i%2==0:
        print(i)
    i+=1
```

```python
# WAP to print odd numbers from 1 to 100 using while loop
i = 1
while i<=100:
    if i%2!=0:
        print(i)
    i+=1
```

```python
import math

number1=12
number2=10

print("\n Mathematical Operations!!!")
print("\n 1. Add")
print("\n 2. Subtract")
print("\n 3. Divide")
print("\n 4. Multiply")
print("\n 5. Modulus")
print("\n 6. Tangential")
print("\n 7. Exponential")
print("\n 8. Natural Log")
print("\n 9. Log10")
print("\n 10. Log2")
print("\n 11. Combination")
print("\n 12. Permutation")
print("\n 13. Factorial")
print("\n e. Factorial")
print("")

while True:
    choice = input("Enter your choice")

    if choice == "1":
        print(f"{number1}+{number2} = {number1+number2}")

    elif choice == "2":
        print(f"{number1}-{number2} = {number1-number2}")
    elif choice == "3":
        print(f"{number1}/{number2} = {number1/number2}")
    elif choice == "4":
        print(f"{number1}*{number2} = {number1*number2}")
    elif choice == "5":
        print(f"{number1}%{number2} = {number1%number2}")
    elif choice == "5":
        print(f"{number1}%{number2} = {number1%number2}")
    elif choice == "6":
        print(f"tan{number1} and tan{number2} = {math.tan(number1)} and {math.tan(number2)} resp.")
    elif choice == "7":
        print(f"exp{number1} and exp{number2} = {math.exp(number1)} and {math.exp(number2)} resp.")
    elif choice == "8":
        print(f"log{number1} and log{number2} = {math.log(number1)} and {math.log(number2)} resp.")
    elif choice == "9":
        print(f"log10{number1} and log10{number2} = {math.log10(number1)} and {math.log10(number2)} resp.")
    elif choice == "10":
        print(f"log2{number1} and log2{number2} = {math.log2(number1)} and {math.log2(number2)} resp.")
    elif choice == "11":
        if number1>number2:
            print(f"comb({number1, number2}) is {math.comb(number1,number2)}")
        else:
            print(f"{number2} is greater than {number1}, so it cannot be executed")
    elif choice == "12":
        if number1>number2:
            print(f"perm({number1, number2}) is {math.perm(number1,number2)}")
        else:
            print(f"{number2} is greater than {number1}, so it cannot be executed")
    elif choice == "13":
        print(f"factorial{number1} and factorial{number2} = {math.factorial(number1)} and {math.factorial(number2)} resp.")
    elif choice == "e":
        print("Program Exitted")
        break
    else:
        print("Invalid Choice")
```
