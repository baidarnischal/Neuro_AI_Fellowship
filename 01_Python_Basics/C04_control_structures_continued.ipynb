```python
# WAP to find the factorial of a number using for loop

number = 6
factorial = 1

for i in range(1,number+1):
    factorial *=i

print(factorial)
```

```python
# WAP to find the factorial of a number using while loop

number = 5
factorial = 1
i = 1

while i<=number:
    factorial*=i
    i+=1

print(factorial)
```

```python
# WAP to reverse a number using while loop

number = 123143
reverse = 0

while number > 0:
    last_digit = number % 10
    reverse = reverse*10 + last_digit
    number = number//10

print(reverse)
    

```

```python
# WAP to reverse a number using for loop

number = 123143
reverse = 0


for i in range(len(str(number))):
    last_digit = number % 10
    reverse = reverse*10 + last_digit
    number = number//10

print(reverse)
    

```

```python
# WAP to find sum of each digit of a number using while loop

number = 123143
sum = 0

while number > 0:
    last_digit = number % 10
    sum+=last_digit
    number = number//10

print(sum)
```

```python
# WAP to print all numbers from 1 to 100 using while loop
i = 1
while i<=100:
    print(i)
    i+=1
```

```python
# WAP to print odd numbers from 1 to 100 using while loop
i = 1
while i<=100:
    if i%2!=0:
        print(i)
    i+=1
```
