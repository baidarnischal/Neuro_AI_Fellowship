# Arithmetic Operators
- `+` Addition
- `-` Subtraction
- `*` Multiplication
- `/` Division
- `//` Floor Division
- `%` Modulus
- `**` Exponentiation

### Exercise on Arithmetic Operators

```python
a = 10
b = 3

print(a + b)   # 13
print(a - b)   # 7
print(a * b)   # 30
print(a / b)   # 3.333...
print(a // b)  # 3
print(a % b)   # 1
print(a ** b)  # 1000
```

```python
# 1. Addition
num_1 = int(input("Enter the first number"))
num_2 = int(input("Enter the second number"))

result = num_1 + num_2
print(f"{num_1} + {num_2} = {result}")
```

```python
# 2. Subtraction
num_1 = int(input("Enter the first number"))
num_2 = int(input("Enter the second number"))

result = num_1 - num_2
print(f"{num_1} - {num_2} = {result}")
```

```python
# 3. Multiplication
num_1 = int(input("Enter the first number"))
num_2 = int(input("Enter the second number"))

result = num_1 * num_2
print(f"{num_1} * {num_2} = {result}")
```

```python
# 4. Division
num_1 = int(input("Enter the first number"))
num_2 = int(input("Enter the second number"))

result = num_1 / num_2
print(f"{num_1} / {num_2} = {result}")
```

```python
# 5. Exponentiation
num_1 = int(input("Enter the first number"))
num_2 = int(input("Enter the second number"))

result = num_1 ** num_2
print(f"{num_1} ** {num_2} = {result}")
```

```python
# 6. Floor Division
num_1 = int(input("Enter the first number"))
num_2 = int(input("Enter the second number"))

result = num_1 // num_2
print(f"{num_1} // {num_2} = {result}")
```

```python
# 7. Modulus
num_1 = int(input("Enter the first number"))
num_2 = int(input("Enter the second number"))

result = num_1 % num_2
print(f"{num_1} % {num_2} = {result}")
```

# Assignment Operators

- `=` Assignment

- `*=` Multiply and assign
- `/=` Divide and assign
- `//=` Floor divide and assign
- `%=` Modulus and assign
- `**=` Exponentiate and assign

```python
# 1. Simple Assignment
var = int(input("Enter a number"))
print(var)
type(var)
```

```python
# 2. Add and Assign
var = int(input("Enter a number"))
print(f"You entered: {var}")

var+=5
print(var)
```

```python
# 3. Subtract and Assign
var = int(input("Enter a number"))
print(f"You entered: {var}")

var-=5
print(var)
```

### Question: Write a program that displays the outputs of AND, OR, NAND, NOR, XOR, and XNOR for all four possible combinations of two binary inputs.

```python
print("A B | AND OR NAND NOR XOR XNOR")
print("--------------------------------")

# Combination 1: A = 0, B = 0
a = 0
b = 0
print(f"{a} {b} |  {a & b}   {a | b}   {int(not (a & b))}    {int(not (a | b))}   {a ^ b}    {int(not (a ^ b))}")

# Combination 2: A = 0, B = 1
a = 0
b = 1
print(f"{a} {b} |  {a & b}   {a | b}   {int(not (a & b))}    {int(not (a | b))}   {a ^ b}    {int(not (a ^ b))}")

# Combination 3: A = 1, B = 0
a = 1
b = 0
print(f"{a} {b} |  {a & b}   {a | b}   {int(not (a & b))}    {int(not (a | b))}   {a ^ b}    {int(not (a ^ b))}")

# Combination 4: A = 1, B = 1
a = 1
b = 1
print(f"{a} {b} |  {a & b}   {a | b}   {int(not (a & b))}    {int(not (a | b))}   {a ^ b}    {int(not (a ^ b))}")
```
